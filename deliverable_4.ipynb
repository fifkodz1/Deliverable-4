## 3. Task 1 — Temperature

The `Temperature` class:

- accepts Celsius (`C`), Fahrenheit (`F`), and Kelvin (`K`) case-insensitively;
- rejects invalid units and values below absolute zero;
- exposes `value` and `unit` as read-only properties;
- converts values between all three supported units;
- returns a new object from `convert_to`;
- provides readable `str` and unambiguous `repr` output;
- implements all six comparison operators across different units;
- uses a tolerance of `1e-9` for floating-point comparisons.

### 3.1 Conversions and representations

```python
room_temperature = Temperature(25, "c")
fahrenheit = room_temperature.convert_to("F")
kelvin = room_temperature.convert_to("K")

print(f"Original:   {room_temperature}")
print(f"Fahrenheit: {fahrenheit}")
print(f"Kelvin:     {kelvin}")
print(f"repr:       {room_temperature!r}")
```

### 3.2 Reference points

The following examples cover absolute zero, the point where Celsius and Fahrenheit are equal, the freezing point of water, normal human body temperature, and the boiling point of water.

```python
reference_points = [
    (-273.15, -459.67, 0.0),
    (-40.0, -40.0, 233.15),
    (0.0, 32.0, 273.15),
    (37.0, 98.6, 310.15),
    (100.0, 212.0, 373.15),
]

print(f"{'Celsius':>10} {'Fahrenheit':>12} {'Kelvin':>10}")
print("-" * 34)
for celsius, expected_fahrenheit, expected_kelvin in reference_points:
    temperature = Temperature(celsius, "C")
    print(
        f"{temperature.to_celsius():10.2f} "
        f"{temperature.to_fahrenheit():12.2f} "
        f"{temperature.to_kelvin():10.2f}"
    )
```

### 3.3 Validation and comparisons

```python
valid_examples = [
    Temperature(0, "C"),
    Temperature(32, "F"),
    Temperature(273.15, "K"),
]

print("Cross-unit equality:", valid_examples[0] == valid_examples[1] == valid_examples[2])
print("Ordering example:", Temperature(0, "C") < Temperature(100, "C"))

for value, unit in [(-274, "C"), (25, "X")]:
    try:
        Temperature(value, unit)
    except ValueError as error:
        print(f"Rejected Temperature({value!r}, {unit!r}): {error}")
```

## 4. Task 2 — Fraction

The `Fraction` class:

- requires integer numerators and denominators;
- requires a strictly positive denominator;
- automatically reduces fractions using the greatest common divisor;
- exposes the numerator and denominator as read-only properties;
- supports addition, subtraction, multiplication, and division;
- supports negation and conversion to `float`;
- implements all six comparison operators;
- provides readable `str` and unambiguous `repr` output.

### 4.1 Reduction and arithmetic

```python
first = Fraction(1, 2)
second = Fraction(1, 3)
reduced = Fraction(8, 12)

print(f"Reduced 8/12: {reduced}")
print(f"{first} + {second} = {first + second}")
print(f"{first} - {second} = {first - second}")
print(f"{first} * {second} = {first * second}")
print(f"{first} / {second} = {first / second}")
print(f"Negative: {-reduced}")
print(f"Float value: {float(Fraction(1, 4))}")
print(f"repr: {reduced!r}")
```

### 4.2 Validation and comparisons

```python
print("Equivalent fractions:", Fraction(1, 2) == Fraction(2, 4))
print("Ordering example:", Fraction(1, 4) < Fraction(3, 4))

invalid_fractions = [(1, 0), (1, -2), (1.5, 2)]
for numerator, denominator in invalid_fractions:
    try:
        Fraction(numerator, denominator)
    except (TypeError, ValueError) as error:
        print(f"Rejected Fraction({numerator!r}, {denominator!r}): {error}")

try:
    Fraction(1, 2) / Fraction(0, 1)
except ZeroDivisionError as error:
    print(f"Division by zero rejected: {error}")
```

## 6. Design decisions

### Temperature
Temperatures are converted to Celsius before comparison, which provides one common scale. A tolerance of `1e-9` is used to account for floating-point rounding. `convert_to` returns a new object instead of modifying the original object. Absolute-zero validation is performed in the unit supplied by the user.

### Fraction
Fractions are reduced during initialization, so every valid object has a consistent representation. A positive denominator simplifies equality and ordering. Arithmetic methods create new reduced `Fraction` objects, while comparisons use exact integer cross-multiplication and therefore avoid floating-point rounding errors.

Both implementations return `NotImplemented` when binary operations or comparisons receive unsupported object types, allowing Python to handle those operations correctly.
